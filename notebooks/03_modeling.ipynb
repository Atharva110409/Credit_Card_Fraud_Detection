# Model Training

In this notebook, we will train and compare three machine learning models:

- Logistic Regression
- Random Forest
- XGBoost

Finally, the trained models will be saved for deployment in the Streamlit application.

```python
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import joblib

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

from sklearn.metrics import accuracy_score
```

```python
X_train = pd.read_csv(
    r"C:\Users\athar\OneDrive\Desktop\credit\data\processed\X_train.csv"
)

X_test = pd.read_csv(
    r"C:\Users\athar\OneDrive\Desktop\credit\data\processed\X_test.csv"
)

y_train = pd.read_csv(
    r"C:\Users\athar\OneDrive\Desktop\credit\data\processed\Y_train.csv"
).squeeze()

y_test = pd.read_csv(
    r"C:\Users\athar\OneDrive\Desktop\credit\data\processed\Y_test.csv"
).squeeze()
```

```python
print("Training Data :", X_train.shape)
print("Testing Data  :", X_test.shape)
print("Training Labels :", y_train.shape)
print("Testing Labels  :", y_test.shape)
```

## Logistic Regression

```python
lr_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

lr_model.fit(X_train, y_train)
```

```python
lr_pred = lr_model.predict(X_test)
```

```python
lr_accuracy = accuracy_score(y_test, lr_pred)

print("Logistic Regression Accuracy :", lr_accuracy)
```

## Random Forest

```python
rf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)
```

```python
rf_pred = rf_model.predict(X_test)
```

```python
rf_accuracy = accuracy_score(y_test, rf_pred)

print("Random Forest Accuracy :", rf_accuracy)
```

## XGBoost

```python
xgb_model = XGBClassifier(
    random_state=42,
    eval_metric="logloss"
)

xgb_model.fit(X_train, y_train)
```

```python
xgb_pred = xgb_model.predict(X_test)
```

```python
xgb_accuracy = accuracy_score(y_test, xgb_pred)

print("XGBoost Accuracy :", xgb_accuracy)
```

## Model Comparison

```python
comparison = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Random Forest",
        "XGBoost"
    ],
    "Accuracy": [
        lr_accuracy,
        rf_accuracy,
        xgb_accuracy
    ]
})

comparison.sort_values(
    by="Accuracy",
    ascending=False
)
```

## Save Trained Models

```python
joblib.dump(
    lr_model,
    r"C:\Users\athar\OneDrive\Desktop\credit\models\logistic_regression.pkl"
)

joblib.dump(
    rf_model,
    r"C:\Users\athar\OneDrive\Desktop\credit\models\random_forest.pkl"
)

joblib.dump(
    xgb_model,
    r"C:\Users\athar\OneDrive\Desktop\credit\models\xgboost.pkl"
)

print("All models saved successfully!")
```
